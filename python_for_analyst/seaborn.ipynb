# 🌊 Statistical Data Visualization with Seaborn

A comprehensive, hands-on guide to advanced exploratory data analysis and statistical visualization using **Seaborn**.

---

### 🎯 What is Seaborn?
**Seaborn** is a powerful Python visualization library built directly on top of **Matplotlib**. It integrates closely with **pandas** data structures and provides a high-level API designed specifically for statistical exploration.

### 🌟 Key Advantages over Pure Matplotlib:
1. **Automated Statistical Aggregation**: Computes means, medians, error bars, and confidence intervals automatically.
2. **Effortless Multi-Dimensional Mapping**: Easily map categories to color (`hue`), marker shapes (`style`), and point sizes (`size`) with a single parameter.
3. **Curated Themes & Palettes**: Beautiful, publication-ready color schemes and aesthetic styles out of the box.
4. **Specialized Statistical Plots**: High-level support for violin plots, swarm plots, joint distribution plots, and correlation heatmaps.
5. **Faceted Multi-Plot Grids**: Create multi-panel subplots split across categorical variables with `col` and `row` parameters without writing nested loops.

## 1. Setup & Environment
Importing Seaborn along with Matplotlib, Pandas, NumPy, and checking package environment.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import os

# Set global figure style for clear inline display
%matplotlib inline
print(f"Seaborn version: {sns.__version__}")
print(f"Pandas version:  {pd.__version__}")
```

The **Palmer Penguins** dataset contains morphological measurements (bill length, bill depth, flipper length, body mass) for three penguin species observed on three islands in Antarctica.

Checking frequency distributions for categorical columns (`species` and `island`).

## 2. Themes, Styles & Color Palettes

Seaborn separates plot configuration into two independent components:
1. **Styles (`set_style`)**: `darkgrid` (default), `whitegrid`, `dark`, `white`, and `ticks`.
2. **Contexts (`set_context`)**: Scales font and line elements for different media: `paper`, `notebook` (default), `talk`, and `poster`.

```python
# Inspect available built-in color palettes
sns.color_palette("Dark2")
```

## 3. Relational Plots: Examining Continuous Relationships

Relational plots allow you to visualize how two or more continuous variables correlate.
- `sns.scatterplot()`: Renders individual observation points on an $(x, y)$ coordinate plane.
- Parameters:
  - `hue`: Encodes a variable with distinct colors.
  - `palette`: Selects the color palette (e.g., `'Dark2'`, `'Set1'`, `'viridis'`).
  - `style`: Encodes a categorical variable with distinct marker shapes.
  - `size`: Encodes a continuous or ordinal variable with marker area size.

```python
# Setting custom aesthetic style and plotting flipper length vs body mass
sns.set_style("dark")
sns.set_context('notebook')

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=penguins,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="island",
    palette="Dark2"
)
plt.title("Flipper Length vs. Body Mass by Island")
plt.show()
```

### Multi-Attribute Scatter Plot
Combining `hue` for island and `style` for biological sex to observe 4 dimensions simultaneously.

```python
plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=penguins,
    x="species",
    y="body_mass_g",
    hue="island",
    palette="Dark2",
    style="sex",
    s=80  # Marker size
)
plt.title("Body Mass across Species, Island, and Sex")
plt.show()
```

### Continuous Size & Transparency Mapping
Mapping `bill_length_mm` to point size (`size`) and adjusting transparency (`alpha=0.7`) to prevent occlusion.

```python
plt.figure(figsize=(9, 5))
sns.set_style("whitegrid")
sns.scatterplot(
    data=penguins,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species",
    size="bill_length_mm",
    sizes=(30, 250),
    alpha=0.7,
    palette="viridis"
)
plt.title("Penguin Morphometrics: Flipper Length vs. Body Mass (Sized by Bill Length)")
plt.show()
```

### Faceted Subplots with `relplot`
Using figure-level `relplot` to split the visualization across multiple columns based on `island` without writing manual Matplotlib loops.

```python
sns.relplot(
    data=penguins,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species",
    col="island",
    kind="scatter",
    palette="Set1",
    height=4,
    aspect=1
)
plt.show()
```

## 4. Distribution Plots: Histograms, KDE & Joint Distributions

Distribution plots allow data analysts to evaluate central tendency, spread, skewness, modality, and outliers.

### Histograms with Kernel Density Estimation (`histplot`)
Grouping continuous values into bins and overlaying a smooth KDE curve.

```python
plt.figure(figsize=(9, 4.5))
sns.histplot(
    data=penguins,
    x="body_mass_g",
    hue="species",
    kde=True,
    bins=25,
    element="step",
    palette="muted"
)
plt.title("Body Mass Distribution by Species with KDE")
plt.xlabel("Body Mass (grams)")
plt.ylabel("Count")
plt.show()
```

### Kernel Density Estimate Plots (`kdeplot`)
Estimating the continuous probability density function (PDF) across groups with filled curves.

```python
plt.figure(figsize=(9, 4.5))
sns.kdeplot(
    data=penguins,
    x="flipper_length_mm",
    hue="species",
    fill=True,
    common_norm=False,
    palette="Dark2",
    alpha=0.4
)
plt.title("Kernel Density Estimation of Flipper Length by Species")
plt.xlabel("Flipper Length (mm)")
plt.ylabel("Density")
plt.show()
```

### Bivariate Joint Distributions (`jointplot`)
Visualizing both the bivariate relationship and the univariate marginal distributions on the top and right margins.

```python
sns.jointplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",
    kind="scatter",
    palette="Dark2",
    height=6
)
plt.show()
```

## 5. Categorical Plots: Comparing Groups & Aggregates

Categorical plots are used when one variable is categorical and the other is numerical.

### 1. Strip Plot (Raw Observations)
Displays all raw individual data points along a category axis. `jitter=True` prevents overlapping points.

```python
plt.figure(figsize=(8, 4.5))
sns.stripplot(
    data=penguins,
    x="species",
    y="body_mass_g",
    hue="island",
    jitter=True,
    palette="Dark2",
    size=6
)
plt.title("Body Mass across Species (Strip Plot with Jitter)")
plt.show()
```

### 2. Swarm Plot (Beeswarm)
Positions points side-by-side along the categorical axis without any overlap, providing a true representation of distribution density and sample size.

```python
plt.figure(figsize=(8, 4.5))
sns.swarmplot(
    data=penguins,
    x="species",
    y="body_mass_g",
    hue="sex",
    palette="Set2",
    size=5
)
plt.title("Body Mass across Species by Sex (Swarm Plot)")
plt.show()
```

### 3. Box Plot (5-Number Summary & Outlier Detection)
Visualizes the median, interquartile range (IQR: Q1 to Q3), whiskers ($1.5 \times IQR$), and outliers.

```python
plt.figure(figsize=(8, 4.5))
sns.boxplot(
    data=penguins,
    x="species",
    y="body_mass_g",
    hue="sex",
    palette="pastel"
)
plt.title("Body Mass Distribution by Species and Sex (Box Plot)")
plt.show()
```

### 4. Violin Plot (KDE + Box Plot Hybrid)
Combines a box plot with a rotated kernel density curve on each side. Splitting violins (`split=True`) enables direct comparison between two binary subgroups (e.g., Male vs. Female).

```python
plt.figure(figsize=(8, 4.5))
sns.violinplot(
    data=penguins,
    x="species",
    y="flipper_length_mm",
    hue="sex",
    split=True,
    inner="quart",
    palette="muted"
)
plt.title("Flipper Length Distribution by Species & Sex (Split Violin Plot)")
plt.show()
```

### 5. Bar Plot (Statistical Aggregator)
Computes point estimates (mean by default) and displays uncertainty error bars (standard deviation with `errorbar="sd"`).

```python
plt.figure(figsize=(8, 4.5))
sns.barplot(
    data=penguins,
    x="species",
    y="body_mass_g",
    hue="species",
    legend=False,
    estimator=np.mean,
    errorbar="sd",
    palette="Blues_d"
)
plt.title("Average Body Mass by Species (with Standard Deviation Error Bars)")
plt.ylabel("Mean Body Mass (g)")
plt.show()
```

### 6. Count Plot (Categorical Frequency)
Specialized bar plot displaying observation counts for each category.

```python
plt.figure(figsize=(8, 4.5))
sns.countplot(
    data=penguins,
    x="species",
    hue="island",
    palette="tab10"
)
plt.title("Count of Penguins per Species and Island")
plt.xlabel("Species")
plt.ylabel("Number of Observations")
plt.show()
```

## 6. Matrix Plots & Correlation Analysis

Correlation heatmaps quantify pairwise linear correlations between continuous numerical features on a scale from $-1.0$ (perfect negative correlation) to $+1.0$ (perfect positive correlation).

```python
# Compute Pearson correlation matrix on numeric features
numeric_penguins = penguins.select_dtypes(include=np.number)
corr_matrix = numeric_penguins.corr()
corr_matrix
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    linewidths=1,
    linecolor="white",
    square=True,
    cbar_kws={"shrink": 0.8}
)
plt.title("Correlation Heatmap of Penguin Morphometrics", fontsize=14, pad=12)
plt.show()
```

## 7. Pairwise Relationships with `pairplot`

`sns.pairplot()` is one of the most powerful exploratory data analysis (EDA) functions in Python. It plots pairwise relationships across an entire DataFrame in a compact grid matrix, displaying scatter plots off-diagonal and univariate distributions (KDE or histograms) along the diagonal. Setting `corner=True` removes duplicate upper-triangle panels for cleaner presentation.

```python
sns.pairplot(
    penguins,
    hue="species",
    palette="Dark2",
    diag_kind="kde",
    corner=True,
    plot_kws={"alpha": 0.7, "s": 35}
)
plt.suptitle("Pairwise EDA Grid across Penguin Measurements", y=1.02, fontsize=14)
plt.show()
```

## 8. Regression & Statistical Relationships

Seaborn makes fitting and plotting linear regression lines simple:
- `sns.regplot()`: Axes-level function for single linear regression with a 95% confidence interval band.
- `sns.lmplot()`: Figure-level function combining `regplot` with FacetGrid to fit separate regressions across categories.

```python
plt.figure(figsize=(8, 5))
sns.regplot(
    data=penguins,
    x="flipper_length_mm",
    y="body_mass_g",
    scatter_kws={"alpha": 0.5, "color": "teal"},
    line_kws={"color": "darkred", "linewidth": 2}
)
plt.title("Linear Regression: Flipper Length vs. Body Mass (with 95% CI)")
plt.xlabel("Flipper Length (mm)")
plt.ylabel("Body Mass (g)")
plt.show()
```

```python
# Fitting separate regression lines across species
sns.lmplot(
    data=penguins,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species",
    palette="Set1",
    aspect=1.4,
    height=5,
    markers=["o", "s", "D"]
)
plt.title("Linear Regression by Species")
plt.show()
```

## 9. Real-World Case Study: Analyzing the Housing Dataset

Let's apply our Seaborn visualization toolkit to the repository's `data/Housing.csv` real estate dataset.

```python
# Load Housing dataset using portable relative path
housing_path = "data/Housing.csv" if os.path.exists("data/Housing.csv") else "Housing.csv"
housing_df = pd.read_csv(housing_path)

print(f"Loaded {len(housing_df)} housing records.")
housing_df.head()
```

### Multi-Panel Housing EDA Dashboard
Examining house price distribution by furnishing status, air conditioning impact, and square footage regression.

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))

# 1. Price comparison by furnishing status & air conditioning
sns.boxplot(
    data=housing_df,
    x="furnishingstatus",
    y="price",
    hue="airconditioning",
    palette="Set2",
    ax=axes[0]
)
axes[0].set_title("House Price by Furnishing Status & Air Conditioning", fontsize=12)
axes[0].set_ylabel("Price ($)")
axes[0].ticklabel_format(style="plain", axis="y")

# 2. Area vs Price regression
sns.regplot(
    data=housing_df,
    x="area",
    y="price",
    scatter_kws={"alpha": 0.4, "color": "#2b5c8f"},
    line_kws={"color": "#d95f02", "linewidth": 2},
    ax=axes[1]
)
axes[1].set_title("Square Footage (Area) vs. Price Regression", fontsize=12)
axes[1].set_xlabel("Area (sq ft)")
axes[1].set_ylabel("Price ($)")
axes[1].ticklabel_format(style="plain", axis="y")

plt.tight_layout()
plt.show()
```

### Housing Numerical Feature Correlation Heatmap

```python
plt.figure(figsize=(8, 6))
housing_numeric = housing_df.select_dtypes(include=np.number)
sns.heatmap(
    housing_numeric.corr(),
    annot=True,
    fmt=".2f",
    cmap="mako",
    linewidths=0.5,
    square=True
)
plt.title("Correlation Heatmap: Housing Numerical Attributes", pad=12)
plt.show()
```

## 10. Summary & Function Reference Table

| Category | Function | Purpose | Best Used For |
| :--- | :--- | :--- | :--- |
| **Relational** | `sns.scatterplot()` | Observation point mapping | Bivariate relationships with hue/size |
| | `sns.lineplot()` | Continuous progression & trends | Time-series or sorted continuous trends with CIs |
| | `sns.relplot()` | Figure-level relational wrapper | Faceted subplots across row/col categories |
| **Distribution** | `sns.histplot()` | Binned value counts & KDE | Distribution spread, skewness, modality |
| | `sns.kdeplot()` | Continuous probability density | Comparing smooth group densities |
| | `sns.jointplot()` | Bivariate + marginals | Deep examination of two continuous variables |
| **Categorical** | `sns.stripplot()` | Jittered raw points | Viewing individual observations per group |
| | `sns.swarmplot()` | Non-overlapping points | Precise distribution density for small/medium data |
| | `sns.boxplot()` | 5-number summary | Spotting outliers and comparing quartiles |
| | `sns.violinplot()` | Boxplot + KDE | Multi-modal distributions across categories |
| | `sns.barplot()` | Group aggregates with error bars | Comparing category means/medians with uncertainty |
| | `sns.countplot()` | Category frequencies | Simple categorical frequency count |
| **Matrix** | `sns.heatmap()` | 2D color-coded matrix | Pairwise correlation matrices and confusion tables |
| **Multi-Plot** | `sns.pairplot()` | Pairwise matrix grid | Quick, comprehensive exploratory data analysis (EDA) |
| **Regression** | `sns.regplot()` / `sns.lmplot()` | Linear fit with confidence intervals | Statistical modeling and trend identification |
