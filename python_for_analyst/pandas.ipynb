# Pandas Fundamentals for Data Analysis

A comprehensive hands-on guide to tabular data manipulation, filtering, cleaning, and exploratory data analysis using **pandas**.

### Topics Covered:
1. **Pandas Series**: Creation, indexing, positional slicing (`iloc`), and label slicing (`loc`)
2. **DataFrame Essentials**: Creation from dictionaries/lists, `.head()`, `.tail()`, `.shape`, `.info()`, `.describe()`
3. **Column Operations & Broadcasting**: Arithmetic calculations, feature creation, and column deletion (`.drop()`)
4. **Logical Filtering**: Conditional querying using `&`, `|`, and boolean masks
5. **Data Cleaning & Missing Data**: Detecting nulls (`isnull()`, `isna()`), imputation (`fillna` with forward/backward fill), and removal (`dropna()`)
6. **Combining Datasets**: Concatenation (`pd.concat`) and relational joins (`pd.merge`)
7. **Real-World Exploratory Data Analysis**: Loading, inspecting, and analyzing `Housing.csv` and `Zomato Restaurant Dataset.csv`

# **Logical Operators**

```python
# and
fru[(fru >10) & (fru <15)]
```

```python
# or
fru[(fru >10) | (fru<10)]
```

```python
# not
fru[~(fru >10)]
```

# **Data Cleaning**

```python
df.isnull().sum()
```

```python
df.dropna()
```

```python
df['Age'].fillna(df['Age'].mean())
```

```python
df['Salary'].fillna(df['Salary'].median())
```

```python
# forward filling (use df.ffill() in modern pandas >=2.1)
df.ffill()
```

```python
# backward filling (use df.bfill() in modern pandas >=2.1)
df.bfill()
```

```python
df
```

```python
# replace specific value
df['Name']=df['Name'].replace("Abel","Yeshi")
df['Name']
```

```python
# duplicated
dup=df[df.duplicated(keep='first')]
dup
```

```python
df.drop_duplicates()
```

```python
df
```

```python
# apply and lamda
def multiply(x):
  return x *2
df['Age'] =df['Age'].apply(multiply)
```

```python
df
```

```python
# lamda to back to original

df['Age']=df['Age'].apply(lambda x:x/2)
```

```python
df
```

```python
location_info ={
  "Department":["IT","HR","Finance"],
  "Location":["Addis Ababa","Bole","Mexico"]

}
```

```python
df2=pd.DataFrame(location_info)
df2
```

```python
pd.concat([df,df2],axis=1)
```

```python
# merge
pd.merge(df,df2,on="Department")
```
