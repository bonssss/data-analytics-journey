# **Data Cleaning**

```python
df.isnull().sum()
```

```python
df.dropna()
```

```python
df['Age'].fillna(df['Age'].mean())
```

```python
df['Salary'].fillna(df['Salary'].median())
```

```python
# forward filling
df.fillna(method="ffill")
```

```python
# backward Filling
df.fillna(method="bfill")
```

```python
df
```

```python
# replace specific value
df['Name']=df['Name'].replace("Abel","Yeshi")
df['Name']
```

```python
# duplicated
dup=df[df.duplicated(keep='first')]
dup
```

```python
df.drop_duplicates()
```

```python
df
```

```python
# apply and lamda
def multiply(x):
  return x *2
df['Age'] =df['Age'].apply(multiply)
```

```python
df
```

```python
# lamda to back to original

df['Age']=df['Age'].apply(lambda x:x/2)
```
